```python
def int_to_excel(n):
    # Convert 0-indexed integer n to an Excel-style column label in lowercase
    n += 1  # Convert to 1-indexed
    result = ""
    while n:
        n, remainder = divmod(n - 1, 26)
        result = chr(97 + remainder) + result  # 97 is ASCII for 'a'
    return result
    
```

```python
def smooth_x(x, y, window=5):
    x = np.asarray(x)
    y = np.asarray(y)
    pad = window // 2
    x_padded = np.pad(x, (pad, pad), mode='edge')
    x_smooth = np.convolve(x_padded, np.ones(window)/window, mode='valid')
    return x_smooth, y
    
```

```python
def interpolate_to_regular_grid(X, Y, U, V, z, dx, method='linear'):
    from scipy.interpolate import RegularGridInterpolator
    x = X[:, 0]
    y = Y[0, :]
    new_x = np.arange(x.min(), x.max(), dx)
    new_y = np.arange(y.min(), y.max(), dx)
    new_z = np.linspace(z.min(), z.max(), len(z))
    rgi_u = RegularGridInterpolator((x, y, z), U, method=method, bounds_error=False, fill_value=np.nan)
    rgi_v = RegularGridInterpolator((x, y, z), V, method=method, bounds_error=False, fill_value=np.nan)
    xx, yy, zz = np.meshgrid(new_x, new_y, new_z, indexing='ij')
    pts = np.stack((xx, yy, zz), axis=-1)
    Un = rgi_u(pts)
    Vn = rgi_v(pts)
    Xn, Yn = np.meshgrid(new_x, new_y, indexing='ij')
    return Xn, Yn, new_z, Un, Vn
    
```

```python
def plot_isosurface(ax, Xn, Yn, dx, zn, Un, Vn, level=-0.2, elev=13, azim=135, flag=False, field=None):
    from mpl_toolkits.mplot3d.art3d import Poly3DCollection
    from skimage.measure import marching_cubes
    from scipy.ndimage import map_coordinates
    ow = calc_ow(Un, Vn, dx*1000, dx*1000, flag=flag)
    sigma_ow = normalize_matrix(ow)

    Xg, Yg, Zg = np.meshgrid(Xn[:, 0], Yn[0, :], zn/1000, indexing='ij')
    if field is None:
        verts, faces, normals, values = marching_cubes(sigma_ow, level=level)
    else:
        verts, faces, normals, values = marching_cubes(field, level=level)
    pts = verts.T
    real_x = map_coordinates(Xg, pts, order=1)
    real_y = map_coordinates(Yg, pts, order=1)
    real_z = map_coordinates(Zg, pts, order=1)
    real_verts = np.vstack((real_x, real_y, real_z)).T

    mesh = Poly3DCollection(real_verts[faces], alpha=0.3)
    mesh.set_edgecolor('k')
    ax.add_collection3d(mesh)
    ax.set_xlim(Xg.min(), Xg.max())
    ax.set_ylim(Yg.min(), Yg.max())
    ax.set_zlim(Zg.min(), Zg.max())
    ax.invert_zaxis()
    ax.set_xlabel('x (km)')
    ax.set_ylabel('y (km)')
    ax.set_zlabel('Depth (km)')
    ax.view_init(elev=elev, azim=azim)
    
```

```python
def compute_vorticity(X, Y, U, V):
    
    dx = np.gradient(X, axis=0)[0,0]
    dy = np.gradient(Y, axis=1)[0,0]
    dzeta = []

    for k in range(U.shape[2]):
        dVdx = np.gradient(V[:, :, k], dx, axis=0)
        dUdy = np.gradient(U[:, :, k], dy, axis=1)
        dzeta.append(dVdx - dUdy)

    return np.stack(dzeta, axis=2)  # shape: (nx, ny, nz)

vort = compute_vorticity(X, Y, U, V)
vort_moc = compute_vorticity(Xn_moc, Yn_moc, Un_moc, Vn_moc)
vort_dop = compute_vorticity(Xn_dop, Yn_dop, Un_dop, Vn_dop)
vort_esp = compute_vorticity(Xn_esp, Yn_esp, Un_esp, Vn_esp)

dic = {
    'ORIG': {'X': X, 'Y': Y, 'z': z, 'U': U, 'V': V, 'W': W_org, 'W_ROMS': W, 'Vort': vort},
    'MOCA': {'X': Xn_moc, 'Y': Yn_moc, 'z': zn_moc/1000, 'U': Un_moc, 'V': Vn_moc, 'W': W_moc, 'Vort': vort_moc, 'df': df_moc, 'c': 'b'},
    'DOPIOE': {'X': Xn_dop, 'Y': Yn_dop, 'z': zn_dop/1000, 'U': Un_dop, 'V': Vn_dop, 'W': W_dop, 'Vort': vort_dop, 'df': df_dop, 'c': 'orangered'},
    'ESPRA': {'X': Xn_esp, 'Y': Yn_esp, 'z': zn_esp/1000, 'U': Un_esp, 'V': Vn_esp, 'W': W_esp, 'Vort': vort_esp, 'df': df_esp, 'c': 'limegreen'},
}
```

```python
from scipy.ndimage import label, generate_binary_structure
def largest_below_threshold(cube, thresh=-0.2):
    mask = cube < thresh
    structure = generate_binary_structure(rank=3, connectivity=1)
    labeled, num_features = label(mask, structure=structure)
    if num_features == 0:
        return np.empty((0, 3), dtype=int)
    counts = np.bincount(labeled.ravel())
    counts = counts[1:]
    largest_label = counts.argmax() + 1
    return np.argwhere(labeled == largest_label)
    
ow_full = calc_ow(U, V, (X[1,0]-X[0,0])*1000, (X[1,0]-X[0,0])*1000, flag=False)
ow = np.ones((ow_full.shape[0], ow_full.shape[1], np.max(np.where(z <= 5))))
for k in range(ow.shape[-1]):
    # ow[:,:,k] = np.where( np.hypot(X, Y) < 75, ow_full[:,:,k], np.nan)
    ow[:,:,k] = ow_full[:,:,k]
sigma_ow = normalize_matrix(ow)
emask = largest_below_threshold(sigma_ow, thresh=-.2)
eddy_sigma_ow = np.full_like(sigma_ow, 0)
for (i, j, k) in zip(emask[:,0], emask[:,1], emask[:,2]):
    eddy_sigma_ow[i, j, k] = sigma_ow[i, j, k]


emask = largest_below_threshold(ow, thresh=-6e-10)
eddy_ow = np.full_like(ow, 0)
for (i, j, k) in zip(emask[:,0], emask[:,1], emask[:,2]):
    eddy_ow[i, j, k] = ow[i, j, k]

k = 3
fig, axs = plt.subplots(1, 2, figsize=(10, 5))

pc0 = axs[0].pcolor(X, Y, eddy_sigma_ow[:, :, k])
axs[0].contour(X, Y, eddy_sigma_ow[:,:,k], levels=[-.2], colors='r')
fig.colorbar(pc0, ax=axs[0])
axs[0].set_title('Eddy Sigma OW')

pc1 = axs[1].pcolor(X, Y, eddy_ow[:, :, k])
axs[1].contour(X, Y, eddy_ow[:,:,k], levels=[-6e-10], colors='r')
fig.colorbar(pc1, ax=axs[1])
axs[1].set_title('OW')

plt.tight_layout()
plt.show()

```
